ambil data review

```python
from google_play_scraper import reviews
import pandas as pd

# Ambil data review
result, _ = reviews(
    'com.shopee.id',
    lang='id',
    country='id',
    count=10000
)

import pandas as pd

df = pd.DataFrame(result)
df.head()
```

ambil kolom penting

```python
df = df[['content', 'score']]
df.rename(columns={'content': 'review', 'score': 'rating'}, inplace=True)
```

# Labeling otomatis

```python
def label_sentimen(rating):
    if rating <= 2:
        return 'negatif'
    elif rating == 3:
        return 'netral'
    else:
        return 'positif'

df['sentimen'] = df['rating'].apply(label_sentimen)
```

# Simpan

```python
df.to_csv('dataset_shopee_playstore.csv', index=False)
```
